# Preprocessing: bad channels and ICA

This is a one-time, per-subject step that has to be done interactively
(a person looking at plots and clicking), so it can't be a plain
script you run once for everyone - it's a notebook you re-run once
per subject.

What it does, and why: eelbrain's automated cleanup steps (filtering,
re-referencing) are the same for every subject, and are already
defined in `experiment.py`. But two things can't be known ahead of
time from code alone:
  - which electrodes were noisy for this particular subject's recording
  - which of the automatically-found ICA components are actually eye
    blinks/movements for this particular subject

Both steps below save their result to that subject's files under
`bids/sub-XX/eeg/` automatically. Once saved, you never redo it -
`analysis/cortical_analysis.ipynb` will just pick the cached result
up the next time it loads that subject.

## How to use this notebook
1. Set SUBJECT in the next cell.
2. Run every cell below it, top to bottom.
3. When each plot opens, inspect it and click to mark/select, then
   close the plot window to continue.
4. Change SUBJECT to the next one and repeat from the top.

Subjects in this dataset: sub-01 through sub-13.

```python
from experiment import e
```

```python
# Change this to the subject you're working on, then run the cells
# below for that subject.
SUBJECT = 'sub-01'
e.set(subject=SUBJECT)
```

## Step 1: mark bad channels

Opens a plot of this subject's raw EEG. Click any electrode that
looks consistently noisy/flat/disconnected to mark it bad, then close
the plot. Saved automatically to
`bids/sub-XX/eeg/sub-XX_task-cocktail-bad_channels.txt`.

```python
e.make_bad_channels()
```

## Step 2: fit ICA and select artifact components

Fits ICA for this subject (if it hasn't been fit already) and opens a
GUI showing each component. Select the ones that are clearly eye
blinks or eye movements (not brain activity), then close the plot.
Saved automatically to `bids/sub-XX/eeg/sub-XX ica-ica.fif`.

```python
e.make_ica_selection(epoch='clean', decim=16)
```

## Next subject

Change `SUBJECT` above to the next one (sub-02, sub-03, ..., sub-13)
and re-run the three cells above for it. Once every subject has been
done, `analysis/cortical_analysis.ipynb` is ready to run.
