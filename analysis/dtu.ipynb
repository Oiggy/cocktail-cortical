# DTU dataset: converging evidence (forward models)

A second, independent two-talker EEG dataset, analysed with the same forward-model (TRF) approach as this project:
Fuglsang, Dau & Hjortkjaer 2017, *Noise-robust cortical tracking of attended speech in real-world acoustic scenes*,
NeuroImage 156:435-444 ([Zenodo 1199011](https://zenodo.org/records/1199011)). That paper decoded attention
(backward models); here the same data go through forward models, to see whether our findings recur.

- 18 subjects, 64-channel BioSemi EEG at 512 Hz
- Two talkers (one male, one female) at +-60 degrees; attend one
- Three rooms: anechoic, mild reverberation, high reverberation; plus single-talker trials

**Everything stays on the cluster** (`dataset/dtu`, next to `dataset/cocktail`); nothing is stored on this computer.
